1.  What is Simple Linear Regression?
-  Simple linear regression is a statistical technique used to examine and model the relationship between two variables by fitting a straight line to observed data. In this method, one variable is treated as the independent variable, which is used to predict or explain the behavior of the dependent variable. The relationship is expressed using an equation of a straight line, where the slope represents how much the dependent variable changes for a one-unit change in the independent variable, and the intercept represents the value of the dependent variable when the independent variable is zero. Simple linear regression helps in understanding trends, making predictions, and analyzing how strongly two variables are related, assuming the relationship between them is linear.  Y=mX+c

-  X → Input / Independent Variable

-  m → Slope

2. What are the key assumptions of Simple Linear Regression?
-  The key assumptions of Simple Linear Regression are that there is a linear relationship between the independent and dependent variables, meaning changes in the predictor lead to proportional changes in the response. The observations are independent of each other, so the value of one data point does not influence another. The errors (residuals) have constant variance, known as homoscedasticity, which means the spread of errors remains the same across all values of the independent variable. Additionally, the errors are normally distributed, especially important for hypothesis testing and confidence intervals, and the mean of the errors is zero, indicating that the model does not systematically overestimate or underestimate the dependent variable.

5.  How do we calculate the slope m in Simple Linear Regression?
-  In simple linear regression, the slope $m$ is determined using the Least Squares Method, which calculates the specific angle that minimizes the sum of the squared vertical distances (residuals) between your observed data points and the regression line.Mathematically, this slope represents the ratio of how the variables change together relative to how the independent variable changes alone; you calculate it by dividing the covariance of $X$ and $Y$ (the sum of the product of their deviations from the mean) by the variance of $X$ (the sum of the squared deviations of $X$ from its mean). Alternatively, if you are working with summary statistics, the slope can be found by multiplying the correlation coefficient ($r$) by the ratio of the standard deviation of $Y$ to the standard deviation of $X$.

6. What is the purpose of the least squares method in Simple Linear Regression?
-  The purpose of the least squares method in Simple Linear Regression is to find the line that best fits the observed data by minimizing the sum of the squared differences between the actual values and the predicted values of the dependent variable. By squaring the errors, the method ensures that positive and negative deviations do not cancel each other out and that larger errors are given more importance. This approach produces the most reliable estimates of the slope and intercept, resulting in a regression line that represents the overall trend in the data as accurately as possible.

7.  How is the coefficient of determination (R²) interpreted in Simple Linear Regression?
-  In Simple Linear Regression, the coefficient of determination (R²) measures how well the independent variable explains the variability in the dependent variable. It represents the proportion of the total variation in the dependent variable that is explained by the regression model. R² ranges from 0 to 1, where a value close to 1 indicates that the model explains most of the variation in the data, and a value close to 0 indicates that the model explains very little. For example, an R² of 0.75 means that 75% of the variation in the dependent variable is explained by the independent variable, while the remaining 25% is due to other factors or random error.

9. What is the main difference between Simple and Multiple Linear Regression?
-  The main difference between Simple Linear Regression and Multiple Linear Regression lies in the number of independent variables used to predict the dependent variable. Simple Linear Regression involves one independent variable to explain or predict the outcome, whereas Multiple Linear Regression involves two or more independent variables. As a result, simple linear regression models the relationship using a single straight line, while multiple linear regression models the combined effect of several predictors on the dependent variable, allowing for a more comprehensive and realistic analysis of complex relationships.

10. What are the key assumptions of Multiple Linear Regression?
-  The key assumptions of Multiple Linear Regression are that there is a linear relationship between the dependent variable and each independent variable, meaning the effect of each predictor on the outcome is linear when other variables are held constant. The observations are independent, so one observation does not influence another. The model assumes homoscedasticity, meaning the errors have constant variance across all levels of the independent variables. The errors are normally distributed, which is important for hypothesis testing and confidence intervals. Additionally, there should be no perfect multicollinearity among the independent variables, meaning the predictors should not be highly correlated with each other, and the mean of the errors is zero, indicating unbiased predictions.

11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
-  Heteroscedasticity refers to a situation in a Multiple Linear Regression model where the variance of the error terms is not constant across all levels of the independent variables. In other words, the spread of residuals changes as the values of the predictors change, rather than remaining uniform. This violates one of the key assumptions of multiple linear regression. While heteroscedasticity does not bias the estimated regression coefficients, it affects the reliability of statistical inference by making standard errors incorrect. As a result, hypothesis tests, confidence intervals, and p-values may be misleading, leading to incorrect conclusions about the significance of independent variables.

12. How can you improve a Multiple Linear Regression model with high multicollinearity?
-  High multicollinearity occurs when independent variables in a Multiple Linear Regression model are highly correlated with each other, which can make coefficient estimates unstable and difficult to interpret. A model can be improved by removing or combining highly correlated variables, since they may be providing redundant information. Another approach is feature selection, keeping only the most relevant predictors based on domain knowledge or statistical criteria. Standardization of variables can help with numerical stability, though it does not eliminate multicollinearity itself. More advanced methods such as principal component analysis (PCA) or regularization techniques like Ridge Regression or Lasso Regression can also be used to reduce the impact of multicollinearity by shrinking or transforming coefficients, leading to more stable and reliable models.

13. What are some common techniques for transforming categorical variables for use in regression models?
-  Categorical variables must be converted into numerical form before they can be used in regression models. One common technique is **dummy coding (one-hot encoding)**, where each category is represented by a binary variable (0 or 1), with one category chosen as a reference group to avoid multicollinearity. Another method is **ordinal encoding**, which assigns ordered numerical values to categories when a natural ranking exists (such as low, medium, high). **Effect coding** is also used in some cases, where categories are coded to compare each group to the overall mean rather than a reference category. For variables with many categories, **target encoding** or **frequency encoding** may be applied, where categories are replaced with statistics such as the mean of the dependent variable or their frequency. These techniques allow categorical data to be effectively incorporated into regression models while preserving meaningful information.


14.  What is the role of interaction terms in Multiple Linear Regression?
-  In Multiple Linear Regression, interaction terms are used to capture situations where the effect of one independent variable on the dependent variable depends on the value of another independent variable. Instead of assuming that each predictor influences the outcome independently, interaction terms allow the model to represent combined or conditional effects. They are created by multiplying two (or more) independent variables together and including this product in the regression equation. Interaction terms help explain more complex real-world relationships, improve model accuracy, and provide deeper insight into how variables work together rather than in isolation.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
-  The interpretation of the **intercept** differs between Simple and Multiple Linear Regression because of the number of independent variables in the model. 18.  What are the limitations of using R² as a sole measure of model performance?
-  Using **R²** as the sole measure of model performance has several limitations:

1. **Does not indicate causation** – A high R² only shows a strong association between variables, but it does not prove that changes in the independent variable cause changes in the dependent variable.
2. **Ignores model bias** – R² does not reveal whether the model is systematically under- or over-predicting; it only measures the proportion of variance explained.
3. **Can be inflated with more variables** – In multiple regression, adding more predictors—even irrelevant ones—can artificially increase R², giving a misleading sense of model quality. Adjusted R² is better for multiple predictors.
4. **Insensitive to overfitting** – A model may have a high R² on training data but perform poorly on new data; R² alone does not assess predictive accuracy.
5. **Does not account for residual patterns** – R² ignores whether assumptions like homoscedasticity or linearity are violated, which can affect the validity of the model.

In short, **R² is useful but should be complemented with other metrics** like adjusted R², RMSE, MAE, residual analysis, and validation on test data for a complete assessment of model performance.


19. How would you interpret a large standard error for a regression coefficient?
-  A **large standard error** for a regression coefficient indicates that the estimate of that coefficient is **highly uncertain or imprecise**. In other words, the coefficient could vary widely if the model were fitted to different samples from the same population. This makes it harder to determine whether the predictor truly has a significant effect on the dependent variable, as reflected by **wide confidence intervals** and potentially **high p-values**. Large standard errors can result from **multicollinearity**, **small sample sizes**, or **high variability in the data**. Practically, a large standard error suggests caution in interpreting that coefficient, and that the corresponding predictor may not reliably explain variation in the dependent variable.


20.  How can heteroscedasticity be identified in residual plots, and why is it important to address it?
-  **Heteroscedasticity** can be identified in residual plots by examining the spread of residuals (errors) across predicted values or an independent variable. In a residual plot:

* If the residuals are **randomly scattered** around zero with roughly constant spread, the assumption of **homoscedasticity** is satisfied.
* If the residuals **fan out or funnel in**—meaning their spread **increases or decreases** as the predicted values change—this indicates **heteroscedasticity**. Patterns like a cone or bow shape are classic signs.

It is important to address heteroscedasticity because it **violates a key regression assumption**. While it does not bias the coefficient estimates, it **makes standard errors unreliable**, which affects **confidence intervals, hypothesis tests, and p-values**, potentially leading to incorrect conclusions about variable significance. Remedies include **transforming the dependent variable**, **weighted least squares**, or using **robust standard errors** to obtain more reliable inference.

23. What is polynomial regression?
-  Polynomial regression is a specialized form of regression analysis used to model relationships where the data follows a non-linear or curved path rather than a straight line. By introducing exponents or "powers" of the independent variable—such as squared ($X^2$) or cubed ($X^3$) terms—into the equation, the model transforms the standard straight regression line into a flexible curve capable of fitting complex shapes like parabolas or waves.While this flexibility allows the model to accurately capture intricate patterns like biological growth rates or projectile motion, it carries a significant risk of overfitting; if the chosen degree of the polynomial is too high, the curve will become overly complex and "wiggly," capturing random noise in the data rather than the true underlying trend.

24. How does polynomial regression differ from linear regression?
-  The fundamental difference lies in the shape of the relationship they model: Linear Regression restricts the fit to a straight line ($Y = mX + c$), assuming that the dependent variable changes at a constant rate, whereas Polynomial Regression models a curved relationship by introducing higher powers of the independent variable (such as $X^2$ or $X^3$).While linear regression is simpler and less prone to over-complicating the data, it will consistently "underfit" complex patterns like growth curves or wave-like fluctuations. In contrast, polynomial regression offers the flexibility to fit these intricate shapes, but this comes with a higher risk of overfitting—creating a curve that is so complex it captures the random noise in the dataset rather than the true underlying trend.

25. When is polynomial regression used?

27. Can polynomial regression be applied to multiple variables?
-  Yes, polynomial regression can be applied to multiple variables, a technique known as Multivariate Polynomial Regression. Instead of fitting a curved line to a single predictor, this method generates a curved 3D surface (or hyper-surface) that captures complex relationships among several inputs simultaneously.Crucially, the regression equation expands to include not only the higher-degree powers of each individual variable (such as $X_1^2$ and $X_2^2$) but also interaction terms (like $X_1 \cdot X_2$), which mathematically account for how one variable's effect depends on the value of another. However, this flexibility comes at a cost: as you add variables, the number of required terms explodes—a problem known as the "Curse of Dimensionality"—making the model computationally heavy and highly prone to overfitting unless you have a large dataset.

28. What are the limitations of polynomial regression?
-  The primary limitation of polynomial regression is its significant tendency toward **overfitting**, especially when high-degree polynomials are used. As the complexity of the model increases, the curve begins to capture random noise in the training data rather than the actual underlying pattern, resulting in a shape that oscillates wildly between points. This leads to poor performance on new, unseen data. Additionally, polynomial regression is notoriously unreliable for **extrapolation**; because polynomial functions inevitably shoot towards positive or negative infinity at their tails, predictions made outside the range of the observed data are often drastically inaccurate. Finally, these models are highly sensitive to **outliers**, where even a single anomalous data point can disproportionately skew the entire curve, making the selection of the correct polynomial degree a critical but difficult challenge.

31. How is polynomial regression implemented in Python?

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
import numpy as np

# Generate some sample data for demonstration
X = np.random.rand(100, 1) * 10 - 5  # 100 samples, 1 feature, values between -5 and 5
y = 0.5 * X**2 + 2 * X + 1 + np.random.randn(100, 1) * 2 # A quadratic relationship with some noise

# Example: 2nd-degree polynomial regression
degree = 2
model = make_pipeline(PolynomialFeatures(degree), LinearRegression())

# X = predictor(s), y = target variable
model.fit(X, y)

# Predict
y_pred = model.predict(X)

print("Model fitted successfully.")
print(f"First 5 predictions: {y_pred[:5].flatten()}")
print(f"First 5 actual values: {y[:5].flatten()}")
```
